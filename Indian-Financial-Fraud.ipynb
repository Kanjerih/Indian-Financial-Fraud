```python
transactions_df['Income_Data_Issue'] = np.where(transactions_df['Annual_Income'] < 10000, 1, 0)

income_median = transactions_df.loc[transactions_df['Income_Data_Issue'] == 0, 'Annual_Income'].median()

transactions_df['Annual_Income'] = np.where(
    transactions_df['Income_Data_Issue'] == 1,
    income_median,
    transactions_df['Annual_Income']
)
```
