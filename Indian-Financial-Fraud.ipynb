## Feature Engineering

Based on EDA findings, the following features were engineered on the full `transactions_df` 
(not the `active_df` subset used during EDA, since the final model needs to handle all transactions).

### 1. Amount_Above_200K
- **Logic:** Flags transactions where `Transaction_Amount > 200,000`.
- **Why:** EDA revealed that no legitimate (non-fraud) transaction in this dataset exceeds ₹200,000 — 
  every transaction above this threshold is fraudulent.
- **Result:** Perfect separator — 100% fraud rate above ₹200K vs. 4.95% below.

### 2. Is_Card_Not_Active
- **Logic:** Flags transactions where `Card_Status` is not 'Active' (i.e., Blocked, Expired, or Lost).
- **Why:** EDA showed that Blocked, Expired, and Lost cards have a 100% fraud rate, while Active 
  cards have a much lower rate (~3.5%). This flag simplifies the raw categorical column into a 
  clean binary signal.
- **Result:** Perfect separator — 100% fraud rate for non-Active cards vs. 3.49% for Active cards.

### 3. Amount_to_Credit_Limit_Ratio
- **Logic:** `Transaction_Amount / Credit_Limit` — the proportion of a customer's credit limit used 
  in a single transaction.
- **Why:** Tests the "credit limit breach" fraud signal described in the dataset documentation. 
  A transaction that consumes a large share of the available limit may indicate fraud.
- **Result:** Meaningful signal — fraud transactions use ~23% of credit limit on average (median 3.7%), 
  vs. ~6% for non-fraud (median 1.1%). Fraud cases can even exceed 100% of the limit (up to 406%), 
  while non-fraud tops out at 277%.

### 4. Is_Late_Night
- **Logic:** Flags transactions occurring between midnight and 5am (`Transaction_Hour` between 0-5).
- **Why:** Tests the "late-night transaction" fraud signal from the dataset description, using a 
  broader time bucket rather than hour-by-hour comparison (which showed only minor variation in EDA).
- **Result:** Weak signal — 5.51% fraud rate late at night vs. 5.35% otherwise. Kept for completeness, 
  but not expected to contribute much predictive power on its own.

### 5. Geo_Mismatch
- **Logic:** Flags transactions where `Customer_State` differs from `Merchant_State`.
- **Why:** Tests whether cross-state transactions (a proxy for geographical anomaly) correlate with 
  fraud, as suggested by the dataset description.
- **Result:** Weak/counterintuitive signal — same-state transactions actually showed a slightly 
  *higher* fraud rate (3.74%) than different-state transactions (3.48%) in the Active-card subset. 
  Likely because cross-state commerce is common and non-suspicious in India, unlike crossing 
  international borders (see Is_International below).

### 6. Income_Data_Issue (created during data cleaning, carried into modeling)
- **Logic:** Flags customers with an implausibly low `Annual_Income` (< ₹10,000), which was then 
  replaced with the dataset median for those rows.
- **Why:** Preserves the fact that these values were data quality issues, in case this flag itself 
  turns out to have predictive value, while ensuring `Annual_Income` itself isn't skewed by 
  unrealistic entries.

### Pre-existing strong feature (not engineered, but confirmed via EDA)
### Is_International
- Already present in the raw dataset. EDA showed international transactions have an 18.3% fraud 
  rate vs. 2.7% for domestic transactions — a ~6.7x difference. One of the strongest standalone 
  predictors after Card_Status and Amount_Above_200K.

### Summary of feature signal strength
| Strength | Features |
|---|---|
| Perfect separator | Card_Status / Is_Card_Not_Active, Amount_Above_200K |
| Strong | Is_International |
| Moderate | Amount_to_Credit_Limit_Ratio |
| Weak | Transaction_Channel, Transaction_Hour, Is_Late_Night, Merchant_Risk_Level, Merchant_Category, Merchant_Rating, Geo_Mismatch, Customer_Segment, Payment_Method |

**Implication for modeling:** Two features act as near-perfect rules, meaning a model should achieve 
very high accuracy on those cases almost trivially. The real modeling challenge lies in detecting 
the remaining "hard" fraud — transactions on Active cards, under ₹200K — where signal is weaker and 
likely depends on feature interactions rather than single variables.

```python
# Step 1: Drop columns that shouldn't go into the model
columns_to_drop = [
    'Transaction_ID', 'Customer_ID', 'Card_ID', 'Merchant_ID',   # ID columns
    'Customer_Name', 'Merchant_Name',                             # Free-text/name columns
    'Fraud_Reason',                                               # Leakage risk
    'Transaction_Status',                                         # Leakage risk - outcome of fraud screening itself
    'Transaction_Date', 'Issue_Date', 'Expiry_Date',              # Raw dates (not directly usable)
    'Customer_Since', 'Merchant_Since',
    'Transaction_Time'                                            # Already extracted as Transaction_Hour
]

model_df = transactions_df.drop(columns=columns_to_drop)

# Step 2: Separate features (X) and target (y)
X = model_df.drop(columns=['Fraud_Flag'])
y = model_df['Fraud_Flag']

# Step 3: Identify categorical columns for encoding
categorical_cols = X.select_dtypes(include='object').columns.tolist()
print(categorical_cols)
```

```python
X[categorical_cols].nunique().sort_values(ascending=False)
```

```python
# Step 4: Drop high-cardinality city columns
X = X.drop(columns=['Customer_City', 'Merchant_City'])

# Update categorical_cols list to remove dropped columns
categorical_cols = [col for col in categorical_cols if col not in ['Customer_City', 'Merchant_City']]

# Step 5: One-hot encode remaining categorical columns
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

# Step 6: Convert boolean columns to int
X = X.astype({col: int for col in X.select_dtypes(include='bool').columns})

# Verify
print(X.shape)
print(X.dtypes.value_counts())
```

```python
X = X.astype({col: int for col in X.select_dtypes(include='bool').columns})
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print(X_train.shape, X_test.shape)
print(y_train.mean(), y_test.mean())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

log_reg = LogisticRegression(class_weight='balanced', random_state=42, max_iter=1000)
log_reg.fit(X_train_scaled, y_train)

y_pred_lr = log_reg.predict(X_test_scaled)

print("Logistic Regression")
print(classification_report(y_test, y_pred_lr))
print(confusion_matrix(y_test, y_pred_lr))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(class_weight='balanced', random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)

print("Random Forest")
print(classification_report(y_test, y_pred_rf))
print(confusion_matrix(y_test, y_pred_rf))
```

```python
from xgboost import XGBClassifier

scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()

xgb_model = XGBClassifier(
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    eval_metric='logloss',
    n_jobs=-1
)
xgb_model.fit(X_train, y_train)

y_pred_xgb = xgb_model.predict(X_test)

print("XGBoost")
print(classification_report(y_test, y_pred_xgb))
print(confusion_matrix(y_test, y_pred_xgb))
```

```python
# Extract feature importances from the retrained XGBoost model
importances = pd.Series(xgb_model.feature_importances_, index=X_train.columns)
importances = importances.sort_values(ascending=False)

# Display the top 15 features
print(importances.head(15))

# Plot the top 15 for a clear visual
plt.figure(figsize=(8,6))
importances.head(15).sort_values().plot(kind='barh', color='#4C72B0')
plt.title('Top 15 Feature Importances (XGBoost) - Leakage-Free')
plt.xlabel('Importance')
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import classification_report

# Get predicted probabilities for the positive class (fraud = 1)
y_proba_xgb = xgb_model.predict_proba(X_test)[:, 1]

# Try a few thresholds
for threshold in [0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2, 0.1]:
    y_pred_threshold = (y_proba_xgb >= threshold).astype(int)
    print(f"--- Threshold: {threshold} ---")
    print(classification_report(y_test, y_pred_threshold))
```

```python
from imblearn.over_sampling import SMOTE

# Step 1: Apply SMOTE only to the training data
smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)

# Step 2: Confirm the new class balance
print("Before SMOTE:", y_train.value_counts())
print("After SMOTE:", y_train_smote.value_counts())

# Step 3: Train XGBoost on the SMOTE-balanced data (no need for scale_pos_weight now, since classes are balanced)
from xgboost import XGBClassifier

xgb_smote = XGBClassifier(random_state=42, eval_metric='logloss', n_jobs=-1)
xgb_smote.fit(X_train_smote, y_train_smote)

# Step 4: Predict probabilities on the (untouched, original) test set
y_proba_smote = xgb_smote.predict_proba(X_test)[:, 1]

# Step 5: Evaluate at default threshold (0.5) first
from sklearn.metrics import classification_report

y_pred_smote_05 = (y_proba_smote >= 0.5).astype(int)
print("--- SMOTE XGBoost, Threshold 0.5 ---")
print(classification_report(y_test, y_pred_smote_05))
```

```python
for threshold in [0.3, 0.4, 0.5, 0.6, 0.7]:
    y_pred_t = (y_proba_smote >= threshold).astype(int)
    print(f"--- SMOTE XGBoost, Threshold {threshold} ---")
    print(classification_report(y_test, y_pred_t))
```

```python
# Fraud Detection Project — Summary Notes

## 1. Dataset Overview
Four relational tables were merged into a single analysis-ready dataset:
- **cards_df** (32,457 rows) — card-level attributes (limit, status, type, network)
- **customers_df** (25,000 rows) — customer demographics and account info
- **merchants_df** (500 rows) — merchant category, risk level, rating
- **transactions_df** (250,000 rows) — the primary transaction-level table

All join keys (`Card_ID`, `Customer_ID`, `Merchant_ID`) were verified unique before merging, 
using left joins to preserve all 250,000 transactions. Final merged dataset: 250,000 rows, 41 columns, 
with zero null values introduced by the merge.

Overall fraud rate: **5.39%** (imbalanced classification problem).

## 2. Data Cleaning
- Converted `Transaction_Date`, `Issue_Date`, `Expiry_Date`, `Customer_Since`, `Merchant_Since` to datetime.
- Converted `Transaction_Time` to a time object and extracted `Transaction_Hour` as a usable numeric feature.
- Fixed a typo in `Marital_Status` ("Divorsed" → "Divorced").
- Identified 968 rows (0.4%) with implausibly low `Annual_Income` (<₹10,000). Flagged these with a new 
  `Income_Data_Issue` column, then replaced the values with the dataset median (calculated from clean 
  rows only, to avoid the outliers skewing the replacement value).
- Verified no duplicate rows and no other typos/inconsistencies across remaining categorical columns.
- Noted 505 future-dated transactions (0.2%) — left as-is, likely a synthetic data artifact, no impact 
  on analysis.

## 3. Exploratory Data Analysis — Key Findings
Fraud rate was examined across every major feature (using an Active-cards-only subset where relevant, 
to avoid the dominant Card_Status effect masking subtler patterns). Signal strength summary:

| Strength | Feature | Finding |
|---|---|---|
| Perfect separator | `Card_Status` | Blocked/Expired/Lost = 100% fraud; Active = 3.49% |
| Perfect separator | `Transaction_Amount` (>₹200K) | No legitimate transaction exceeds ₹200,000 |
| Strong | `Is_International` | 18.3% fraud (international) vs 2.7% (domestic) — 6.7x difference |
| Moderate | `Amount_to_Credit_Limit_Ratio` | Fraud uses ~23% of credit limit on average vs ~6% for non-fraud |
| Weak | `Transaction_Channel`, `Transaction_Hour`, `Merchant_Risk_Level`, `Merchant_Category`, `Merchant_Rating`, `Geo_Mismatch` (state-level), `Customer_Segment`, `Payment_Method` | All showed <1 percentage point variation in fraud rate |

Notably, several features the dataset description flagged as "key fraud signals" (merchant risk level, 
time-of-day, channel) showed only weak standalone signal — suggesting fraud in this dataset is driven 
by a small number of dominant rules plus subtler feature interactions, rather than many independently 
strong predictors.

**Residual fraud check:** after removing non-Active cards and transactions >₹200K, the remaining fraud 
rate among -normal-looking- transactions was still 3.04% — confirming a genuinely harder subset of 
fraud exists beyond the two obvious rules.

## 4. Feature Engineering
New features created based on EDA findings:
- `Amount_Above_200K` — binary flag for the ₹200K non-fraud ceiling
- `Is_Card_Not_Active` — binary flag simplifying Card_Status into a clean predictor
- `Amount_to_Credit_Limit_Ratio` — Transaction_Amount / Credit_Limit
- `Is_Late_Night` — binary flag for transactions between midnight and 5am (weak signal, kept anyway)
- `Geo_Mismatch` — binary flag for Customer_State ≠ Merchant_State (weak/counterintuitive signal)

## 5. Data Leakage Discovery — Important Lesson
An initial model run showed `Transaction_Status` as the dominant feature by a wide margin (importance 
0.455, ~9x the next feature). Investigation revealed this was **data leakage**: `Transaction_Status` 
(Successful/Failed/Declined) is an *outcome* of a transaction being screened for fraud, not information 
available beforehand. Its fraud rates (Declined: 59.6%, Failed: 31.2%, Successful: 1.4%) confirmed it 
was essentially encoding the answer.

**Action taken:** `Transaction_Status` was dropped and all models were retrained. This is a key 
methodological lesson — feature importance review is essential for catching leakage that EDA groupby 
checks alone may miss, since leakage often only becomes obvious once a model over-relies on a feature.

## 6. Modeling — Data Preparation
- Dropped ID columns, free-text columns, raw date columns, and the leaky `Transaction_Status` column.
- Dropped high-cardinality `Customer_City`/`Merchant_City` (85 unique values each) — state-level 
  geography plus `Is_International`/`Geo_Mismatch` already capture relevant location signal.
- One-hot encoded remaining 19 categorical columns (all ≤15 categories) using `drop_first=True` to 
  avoid multicollinearity.
- Converted resulting boolean columns to integers for compatibility across all ML libraries.
- Final feature set: 100 columns, 250,000 rows.
- Split into train/test (80/20) using stratified sampling to preserve the 5.39% fraud ratio in both sets.

## 7. Model Comparison (Leakage-Free, Default Threshold 0.5)

| Model | Precision (fraud) | Recall (fraud) | F1-score (fraud) |
|---|---|---|---|
| Logistic Regression (class_weight='balanced') | 0.40 | 0.71 | 0.51 |
| Random Forest (class_weight='balanced') | 0.99 | 0.51 | 0.67 |
| XGBoost (scale_pos_weight) | 0.49 | 0.75 | 0.59 |

Feature importance from XGBoost confirmed the model relies on legitimate, real-world-available 
signals: `Is_Card_Not_Active` (dominant), `Transaction_Amount`, `Is_International`, and various 
`Merchant_Category` splits — consistent with EDA findings, with no further leakage detected.

## 8. Threshold Tuning
Default 0.5 threshold is rarely optimal for imbalanced classification. A sweep across thresholds 
0.1–0.9 on the XGBoost model found:

| Threshold | Precision | Recall | F1-score |
|---|---|---|---|
| 0.5 (default) | 0.49 | 0.75 | 0.59 |
| 0.7 | 0.71 | 0.70 | 0.70 |
| **0.8** | **0.82** | **0.67** | **0.74** ⭐ |
| 0.9 | 0.92 | 0.62 | 0.74 |

**Threshold 0.8 selected as optimal** — matches the peak F1-score of 0.9 but with meaningfully 
better recall (67% vs 62%), catching more real fraud for the same overall balance.

## 9. SMOTE Comparison
Applied SMOTE to the training data only (never the test set, to avoid leakage) to balance the 
10,778 fraud cases up to match the 189,222 non-fraud cases. Re-trained XGBoost without 
`scale_pos_weight` (redundant once classes are balanced) and swept thresholds 0.3–0.7.

Best SMOTE result: **F1-score 0.71** (thresholds 0.4–0.5) — underperforming the `scale_pos_weight` 
+ threshold-tuning approach (F1 0.74). 

**Conclusion:** for this dataset, class reweighting (`scale_pos_weight`) combined with threshold 
tuning outperformed SMOTE. This is a useful finding — more sophisticated resampling isn't always 
better; it depends on the underlying data structure. Given several near-perfect separator features 
exist in this dataset, direct class reweighting likely preserved that signal better than SMOTE's 
synthetic interpolation between fraud examples.

## 10. Final Model
**XGBoost, trained with `scale_pos_weight`, evaluated at a decision threshold of 0.8:**
- Precision (fraud): 0.82
- Recall (fraud): 0.67
- F1-score (fraud): 0.74

This model catches roughly two-thirds of all fraud cases while keeping false alarms low enough 
(82% of flagged transactions are genuinely fraudulent) to be operationally practical for manual 
review workflows.

## Key Takeaways
1. Always check feature importance for leakage — a feature that dominates suspiciously (like 
   `Transaction_Status` here) is a red flag worth investigating before trusting model results.
2. Default classification thresholds (0.5) are rarely optimal for imbalanced problems — systematic 
   threshold tuning improved F1-score from 0.59 to 0.74 with no retraining required.
3. More complex imbalance-handling techniques (SMOTE) aren't automatically better than simpler 
   approaches (class reweighting) — both should be tested and compared empirically.
4. A small number of dominant features (`Card_Status`, transaction amount ceiling) explain much of 
   the fraud in this dataset, but a genuinely harder residual fraud problem (~3% rate) remains 
   within "normal-looking" transactions — this is where model quality matters most.
```
