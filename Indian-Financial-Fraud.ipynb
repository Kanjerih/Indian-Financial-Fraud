```python
import matplotlib.pyplot as plt

# Chart 1: Fraud rate by Card_Status (use full transactions_df, not active_df)
fraud_by_card_status = transactions_df.groupby('Card_Status')['Fraud_Flag'].mean().sort_values(ascending=False)

plt.figure(figsize=(6,4))
fraud_by_card_status.plot(kind='bar', color='#4C72B0')
plt.title('Fraud Rate by Card Status')
plt.ylabel('Fraud Rate')
plt.xlabel('Card Status')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# 1. Amount_Above_200K — flags transactions above the ₹200K non-fraud ceiling
transactions_df['Amount_Above_200K'] = np.where(transactions_df['Transaction_Amount'] > 200000, 1, 0)

# 2. Is_Card_Not_Active — flags any card status other than 'Active'
transactions_df['Is_Card_Not_Active'] = np.where(transactions_df['Card_Status'] != 'Active', 1, 0)

# 3. Amount_to_Credit_Limit_Ratio — how much of the credit limit this transaction uses
transactions_df['Amount_to_Credit_Limit_Ratio'] = transactions_df['Transaction_Amount'] / transactions_df['Credit_Limit']

# 4. Is_Late_Night — flags transactions between midnight and 5am (inclusive)
transactions_df['Is_Late_Night'] = np.where(transactions_df['Transaction_Hour'].between(0, 5), 1, 0)

# 5. Geo_Mismatch — carry forward from earlier (built on active_df, now applied to full transactions_df)
transactions_df['Geo_Mismatch'] = np.where(transactions_df['Customer_State'] != transactions_df['Merchant_State'], 1, 0)

# Income_Data_Issue already exists in transactions_df from earlier cleaning step — no need to recreate
```
